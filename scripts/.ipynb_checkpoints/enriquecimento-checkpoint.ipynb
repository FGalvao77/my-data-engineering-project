```python
import duckdb
```

```python
con = duckdb.connect(
    database='data_duckdb.db',
    read_only=False
)
```

```python
con.execute(
    '''
    SELECT *
    FROM bronze_z0019
    '''
)
```

```python
df = con.execute(
    '''
    SELECT *
    FROM bronze_z0019
    '''
).fetchdf()

df
```

```python
df = con.execute(
    '''
    SELECT *
    FROM bronze_z0019
    WHERE date_intake >= '2025-06-24'
    '''
).fetchdf()

df
```

```python
df = con.execute(
    '''
    SELECT *, ROW_NUMBER() OVER (PARTITION BY NATBR ORDER BY date_intake DESC) AS row
    FROM bronze_z0019
    WHERE date_intake >= '2025-06-24'
    '''
).fetchdf()

df
```

```python
df = con.execute(
    '''
    SELECT *
    FROM (
        SELECT *, ROW_NUMBER() OVER (PARTITION BY NATBR ORDER BY date_intake DESC) AS row
        FROM bronze_z0019
        WHERE date_intake >= '2025-06-24'
    )
    '''
).fetchdf()

df
```

```python
df = con.execute(
    '''
    SELECT *
    FROM (
        SELECT *, ROW_NUMBER() OVER (PARTITION BY NATBR ORDER BY date_intake DESC) AS row
        FROM bronze_z0019
        WHERE date_intake >= '2025-06-24'
    ) WHERE row = 1
    '''
).fetchdf()

df
```

```python
df
```

```python
df_final = df.drop(columns=['file_name', 'date_intake', 'row'])

df_final
```

```python
df_final.rename(columns={'NATBR':'id'})
```

```python
df_final = df_final.rename(columns={'NATBR':'id',            # identificador
                                    'MAKTX': 'product_name', # nome do produto 
                                    'WERKS': 'id_category',     # categória
                                    'MAINS': 'id_supplier',     # fornecedor
                                    'LABST': 'value_price'})       # preço

df_final
```
