```python
import duckdb
```

```python
con = duckdb.connect(
    database='data_duckdb.db',
    read_only=False
)
```

```python
con.execute(
    '''
    SELECT *
    FROM bronze_z0019
    '''
)
```

```python
df = con.execute(
    '''
    SELECT *
    FROM bronze_z0019
    '''
).fetchdf()

df
```

```python
df = con.execute(
    '''
    SELECT *
    FROM bronze_z0019
    WHERE date_intake >= '2025-06-24'
    '''
).fetchdf()

df
```

```python
df = con.execute(
    '''
    SELECT *, ROW_NUMBER() OVER (PARTITION BY NATBR ORDER BY date_intake DESC) AS row
    FROM bronze_z0019
    WHERE date_intake >= '2025-06-24'
    '''
).fetchdf()

df
```

```python
df = con.execute(
    '''
    SELECT *
    FROM (
        SELECT *, ROW_NUMBER() OVER (PARTITION BY NATBR ORDER BY date_intake DESC) AS row
        FROM bronze_z0019
        WHERE date_intake >= '2025-06-24'
    )
    '''
).fetchdf()

df
```

```python
df = con.execute(
    '''
    SELECT *
    FROM (
        SELECT *, ROW_NUMBER() OVER (PARTITION BY NATBR ORDER BY date_intake DESC) AS row
        FROM bronze_z0019
        WHERE date_intake >= '2025-06-24'
    ) WHERE row = 1
    '''
).fetchdf()

df
```

```python
df
```

```python
df_final = df.drop(columns=['file_name', 'date_intake', 'row'])

df_final
```

```python
df_final.rename(columns={'NATBR':'id'})
```

```python
df_final = df_final.rename(columns={'NATBR':'id',            # identificador
                                    'MAKTX': 'product_name', # nome do produto 
                                    'WERKS': 'id_category',     # categória
                                    'MAINS': 'id_supplier',     # fornecedor
                                    'LABST': 'value_price'})       # preço

df_final
```

```python
df_final.dtypes
```

```python
df2 = df_final.copy()

df2
```

```python
df2 = df2.astype(
    {
        'id': 'int32',
        'product_name': 'string',
        'id_category': 'string',
        'id_supplier': 'int32',
        'value_price': 'float32'
    }
)
```

```python
df2.info()
```

```python
df2.dtypes
```

```python
df2 = df2.astype(
    {
        'id': int,
        'product_name': str,
        'id_category': str,
        'id_supplier': int,
        'value_price': float
    }
)
```

```python
df2.dtypes
```

```python
con.execute(
    '''
    CREATE TABLE IF NOT EXISTS products (
        id BIGINT,
        product_name TEXT,
        id_category TEXT,
        id_supplier BIGINT,
        value_price FLOAT
    )
    '''
)
```

```python
df2
```

```python
df_result = con.execute(
    '''
    SELECT *
    FROM products
    '''
).fetchdf()

df_result
```

```python
con.execute(
    '''
    INSERT INTO products
    SELECT *
    FROM df2
    '''
)
```
