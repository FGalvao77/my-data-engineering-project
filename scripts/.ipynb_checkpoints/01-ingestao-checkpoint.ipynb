```python
file_name = 'z0019_1.csv'
date_intake = datetime.now()

df = pd.read_csv(
    filepath_or_buffer=f'../landing/{file_name}',
    sep=';',
    header=0
)

df['file_name'] = file_name
df['date_intake'] = date_intake

df.head()
```
